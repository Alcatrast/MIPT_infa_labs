# Задача 1

Сгенерируйте 500 выборок из 2500 случайных значений для

a) Экспоненциального распределения (loc=0, scale=13)

б) Бернулевского распределения (p = 0.75)

Постройте график зависимость $S_k$ от $k$, где
$$ S_k =  \frac{1}{k} \sum_{j=1}^{k} ξ_j$$
а $k$ - размер

Отобразите на обоих графиках среднее и напишите небольшой вывод.

```python
# Параметры
n_samples = 500
n_values = 2500

exp_data = np.random.exponential(scale=13, size=(n_samples, n_values))
bern_data = np.random.binomial(n=1, p=0.75, size=(n_samples, n_values))

k_values = np.arange(1, n_values + 1)
exp_S_k = np.cumsum(exp_data, axis=1) / k_values
bern_S_k = np.cumsum(bern_data, axis=1) / k_values

exp_mean_k = np.mean(exp_S_k, axis=0)
bern_mean_k = np.mean(bern_S_k, axis=0)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(k_values, exp_mean_k, label='Эмпирическое среднее')
axes[0].axhline(13, color='red', linestyle='--', label='Теоретическое среднее (13)')
axes[0].set_xlabel('Размер выборки k')
axes[0].set_ylabel('S_k')
axes[0].set_title('Экспоненциальное распределение')
axes[0].legend()

axes[1].plot(k_values, bern_mean_k, label='Эмпирическое среднее', color='green')
axes[1].axhline(0.75, color='red', linestyle='--', label='Теоретическое среднее (0.75)')
axes[1].set_xlabel('Размер выборки k')
axes[1].set_ylabel('S_k')
axes[1].set_title('Распределение Бернулли')
axes[1].legend()

plt.tight_layout()
plt.show()

# Вывод по Задаче 1:
# Построенные графики наглядно иллюстрируют закон больших чисел.
# Для обоих распределений (экспоненциального и Бернулли) эмпирическое 
# бегущее среднее S_k сходится к теоретическому математическому ожиданию 
# (13.0 и 0.75 соответственно) при увеличении размера выборки k.
# 
# При малых значениях k наблюдаются значительные случайные отклонения, 
# которые постепенно затухают. Скорость сходимости обратно пропорциональна 
# дисперсии распределения: для распределения Бернулли (D = p(1-p) = 0.1875) 
# стабилизация наступает быстрее, чем для экспоненциального (D = scale² = 169), 
# что полностью согласуется с теоретическими предсказаниями.
```

# Задача 2

Сгенерируйте 100 выборок из 1000 случайных значений для

a) Экспоненциального распределения (loc=0, scale=13)

б) Распределения Пуассона(mu = 5)

Постройте гистограммы среднего случайных значений для 5ти, 10ти, 50ти, 100 выборок

Поверх каждой нарисуйте ядерную оценку плотности(`sns.kdeplot`) и нормальное распределение с параметрами парамерами нормального распределения к которому они сходятся

Напишите свои выводы

```python
n_samples_2 = 100
n_values_2 = 1000

exp_data_2 = np.random.exponential(scale=13, size=(n_samples_2, n_values_2))
pois_data_2 = np.random.poisson(lam=5, size=(n_samples_2, n_values_2))

sizes = [5, 10, 50, 100]

fig, axes = plt.subplots(4, 2, figsize=(14, 16))

for i, n in enumerate(sizes):
    exp_means = np.mean(exp_data_2[:, :n], axis=1)
    pois_means = np.mean(pois_data_2[:, :n], axis=1)
    
    ax_exp = axes[i, 0]
    
    sns.histplot(exp_means, kde=True, stat="density", color='blue', alpha=0.5, 
                 label='Эмпирическая плотность (KDE)', ax=ax_exp)
    
    x_exp = np.linspace(ax_exp.get_xlim()[0], ax_exp.get_xlim()[1], 100)
    
    ax_exp.plot(x_exp, sps.norm.pdf(x_exp, loc=13, scale=13/np.sqrt(n)), 
                'b--', linewidth=2, label=f'Теор. норм. $N(13, 13^2/{n})$')
    
    ax_exp.set_title(f'Экспоненциальное распределение, n = {n}')
    ax_exp.set_xlabel('Значение выборочного среднего')
    ax_exp.set_ylabel('Плотность')
    ax_exp.legend()
    
    ax_pois = axes[i, 1]
    
    sns.histplot(pois_means, kde=True, stat="density", color='orange', alpha=0.5, 
                 label='Эмпирическая плотность (KDE)', ax=ax_pois)
    
    x_pois = np.linspace(ax_pois.get_xlim()[0], ax_pois.get_xlim()[1], 100)
    
    ax_pois.plot(x_pois, sps.norm.pdf(x_pois, loc=5, scale=np.sqrt(5)/np.sqrt(n)), 
                 'r--', linewidth=2, label=f'Теор. норм. $N(5, 5/{n})$')
    
    ax_pois.set_title(f'Распределение Пуассона, n = {n}')
    ax_pois.set_xlabel('Значение выборочного среднего')
    ax_pois.set_ylabel('Плотность')
    ax_pois.legend()

plt.suptitle('Демонстрация Центральной предельной теоремы', fontsize=16, y=1.01)
plt.tight_layout()
plt.show()

# Вывод:
#
#Центральная предельная теорема наглядно подтверждается:
#независимо от формы исходного распределения (экспоненциального или Пуассона),
#распределение выборочного среднего стремится к нормальному при увеличении n.
#
#Скорость сходимости зависит от исходного распределения:
#для сильно скошенного экспоненциального распределения требуется больший
#размер выборки для достижения нормальности, чем для менее скошенного
#распределения Пуассона с μ = 5.
```

# Задача 3

Сгенерируйте 10 выборок из 1000 случайных значений для

a) Экспоненциального распределения (loc=0, scale=13)

б) Распределения Пуассона(mu = 5)

Постройте графики среднего и медианы в зависимости от числа элементов выорки

Поверх каждой нарисуйте заданное среднее

Напишите свои выводы

```python
n_samples_3 = 10
n_values_3 = 1000

exp_data_3 = np.random.exponential(scale=13, size=(n_samples_3, n_values_3))
pois_data_3 = np.random.poisson(lam=5, size=(n_samples_3, n_values_3))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for dist_name, data, theo_mean in zip(['Экспоненциальное', 'Пуассон'], 
                                      [exp_data_3, pois_data_3], 
                                      [13, 5]):
    ax = axes[0] if dist_name == 'Экспоненциальное' else axes[1]

    cumsum_data = np.cumsum(data, axis=1)
    k_vals = np.arange(1, n_values_3 + 1)
    means = cumsum_data / k_vals  # форма (10, 1000)
    
    medians = np.zeros((n_samples_3, n_values_3))
    for i in range(n_samples_3):
        for k in range(1, n_values_3 + 1):
            medians[i, k-1] = np.median(data[i, :k])
    
    mean_k = np.mean(means, axis=0)
    median_k = np.mean(medians, axis=0)
    theo_median = 13 * np.log(2) if dist_name == 'Экспоненциальное' else 5 

    ax.plot(k_vals, mean_k, label='Среднее')
    ax.plot(k_vals, median_k, label='Медиана')
    ax.axhline(theo_mean, color='red', linestyle='--', label=f'Теор. среднее ({theo_mean})')
    ax.axhline(theo_median, color='green', linestyle=':', label=f'Теор. медиана ({theo_median:.2f})')
    
    ax.set_title(dist_name)
    ax.set_xlabel('Размер выборки k')
    ax.set_ylabel('Значение')
    ax.legend()

plt.tight_layout()
plt.show()

#Вывод:
#Закон больших чисел: и выборочное среднее, и выборочная медиана сходятся к своим теоретическим значениям при росте размера выборки k.
#Влияние асимметрии распределения:
#Для экспоненциального распределения среднее и медиана сходятся
#к разным предельным значениям (13 и 9 соответственно).
#наглядно демонстрирует правостороннюю асимметрию распределения.
#Для распределения Пуассона (mu=5) среднее и медиана сходятся к одному
#значению 5, так как при mu=5 распределение уже достаточно симметрично.
#
#Скорость сходимости:
#Медиана сходится медленнее среднего и обладает большей дисперсией на малых
#выборках, что особенно заметно для асимметричного экспоненциального
#распределения (сильные колебания оранжевой линии в начале графика).
#Среднее является более устойчивой оценкой на малых выборках.
```

# Задача 4

Реализуйте `SGD` и найдите минимум функции `MSE`(lr = 1e-4, iter = 10000, seed = 42, batch_size= 256, нулевое начальное приближение) для следующих данных

Нарисуйте график зависимости лосса от итерации.

Напишите вывод

```python
data = np.load('hw_final_open_data.npy', allow_pickle=False)
target = np.load('hw_final_open_target.npy', allow_pickle=False)

X = np.hstack([np.ones((data.shape[0], 1)), data])
y = target

lr = 1e-4
n_iter = 10000
seed = 42
batch_size = 256

np.random.seed(seed)

n_features = X.shape[1]
w = np.zeros(n_features)

n_samples = X.shape[0]
loss_history = []

for i in range(n_iter):
    indices = np.random.choice(n_samples, batch_size, replace=False)
    X_batch = X[indices]
    y_batch = y[indices]
    
    y_pred = X_batch @ w
    error = y_pred - y_batch
    grad = (2 / batch_size) * (X_batch.T @ error)
    w = w - lr * grad
    
    if i % 100 == 0 or i == n_iter - 1:
        full_y_pred = X @ w
        full_mse = np.mean((full_y_pred - y) ** 2) 
        loss_history.append((i, full_mse))

iterations = [item[0] for item in loss_history]
losses = [item[1] for item in loss_history]

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(iterations, losses, color='blue', linewidth=2, alpha=0.8, label='MSE (на всей выборке)')
ax.set_title('Зависимость лосса (MSE) от итерации при обучении SGD', fontsize=14, pad=15)
ax.set_xlabel('Итерация', fontsize=12)
ax.set_ylabel('Значение MSE', fontsize=12)

ax.legend()
plt.tight_layout()
plt.show()

print(f"Итоговое значение MSE: {losses[-1]:.6f}")
# Вывод: График показывает монотонное убывание функции потерь MSE, 
# что подтверждает корректность работы алгоритма стохастического градиентного спуска. 
```
